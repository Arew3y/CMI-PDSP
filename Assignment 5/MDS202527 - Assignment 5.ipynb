```python
class Node:
    def __init__(self, v=None):
        # start as a single leaf
        self.value = v
        self.left = None
        self.right = None
        self.tag = "L"

    def find(self, v: int):
        # search for v in the tree
        if self.value is None:
            return False
        elif self.tag == "I":
            if v < self.value:
                return self.left.find(v)
            else:
                return self.right.find(v)
        elif self.value == v and self.tag == "L":
            return True
        else:
            return False

    def insert(self, v: int):
        # insert v as a new leaf, using the special rules
        if self.value is None:
            self.value = v
            return None
        elif self.tag == "I":
            if v < self.value:
                return self.left.insert(v)
            else:
                return self.right.insert(v)
        elif self.tag == "L":
            # split this leaf into an internal node with 2 leaves
            if v < self.value:
                self.left = Node(v)
                self.right = Node(self.value)
                self.tag = "I"
                return None
            else:
                self.left = Node(self.value)
                self.right = Node(v)
                self.value = v
                self.tag = "I"
                return None
        else:
            raise ValueError(f"Invalid tag: {self.tag}.")

    def deletea(self, v, intnode = None, prev = None):
        # recursive helper for delete
        if self.tag == "I":
            # go down to the correct child
            if v < self.value:
                self.left.deletea(v, None, self)
            else:
                self.right.deletea(v, None, self)

            # if this node is still internal, fix its value
            if self.tag == "I":
                # internal node value = minimum leaf in right subtree
                n = self.right
                while n.tag == "I":
                    n = n.left
                self.value = n.value
            return None

        elif self.tag == "L":
            # we are at a leaf now
            if self.value != v:
                # this leaf does not contain v
                return None

            # if this is the only node (root leaf), just clear it
            if prev is None:
                self.value = None
                return None

            # otherwise remove this leaf and promote its sibling
            if prev.left is self:
                sibling = prev.right
            else:
                sibling = prev.left

            prev.tag = sibling.tag
            prev.value = sibling.value
            prev.left = sibling.left
            prev.right = sibling.right
            return None

        else:
            return None

    def delete(self, v):
        # public delete function
        return self.deletea(v)

    def strarr(self):
        # inorder traversal, collect (tag, value) pairs
        if self.tag == "I":
            return self.left.strarr() + [(self.tag,self.value)] + self.right.strarr()
        else:
            return [(self.tag,self.value)]

    def __str__(self):
        # print the inorder list so it is easy to check
        return str(self.strarr())
```
